```python
# prompt: Generate a python code for receipe of 20 chocolate brownie of 100gm each.

def chocolate_brownies(num_brownies=20, brownie_weight_grams=100):
    """
    Generates a recipe for chocolate brownies.

    Args:
      num_brownies: The number of brownies to make.
      brownie_weight_grams: The weight of each brownie in grams.
    """

    total_weight = num_brownies * brownie_weight_grams

    # Adjust ingredients based on desired number and weight of brownies
    ingredients = {
        "Unsalted butter": f"{total_weight * 0.25 / 100:.2f} g", # 25% of total weight
        "Unsweetened chocolate": f"{total_weight * 0.2 / 100:.2f} g", #20% of total weight
        "Granulated sugar": f"{total_weight * 0.3 / 100:.2f} g", # 30% of total weight
        "Light brown sugar": f"{total_weight * 0.15 / 100:.2f} g", #15% of total weight
        "Large eggs": f"{num_brownies * 0.5:.0f}", #Half an egg per brownie
        "Vanilla extract": f"{num_brownies * 0.01:.2f} tsp", # 0.01 tsp per brownie
        "All-purpose flour": f"{total_weight * 0.10 / 100:.2f} g", # 10% of total weight
        "Unsweetened cocoa powder": f"{total_weight * 0.05 / 100:.2f} g", # 5% of total weight
        "Salt": f"{num_brownies * 0.005:.2f} tsp", # 0.005 tsp per brownie
    }

    print("Recipe for", num_brownies, "chocolate brownies (", brownie_weight_grams, "g each):")
    for ingredient, quantity in ingredients.items():
        print(f"- {ingredient}: {quantity}")
    print("\nInstructions (example, adjust as needed):")
    print("1. Preheat oven to 350°F (175°C).")
    print("2. Melt butter and chocolate together.")
    print("3. Mix in sugars, eggs, and vanilla.")
    print("4. Combine dry ingredients (flour, cocoa powder, salt).")
    print("5. Gradually add dry ingredients to wet ingredients.")
    print("6. Pour batter into a greased baking pan.")
    print("7. Bake for 25-30 minutes, or until a toothpick inserted comes out with moist crumbs.")
    print("8. Let cool completely before cutting and serving.")

# Example usage for 20 brownies of 100g each
chocolate_brownies()
```
